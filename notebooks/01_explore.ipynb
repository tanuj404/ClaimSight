```python
print("Missing Insurance_Type only:", df["Insurance_Type"].isnull().sum())
print("Missing Provider_Specialty only:", df["Provider_Specialty"].isnull().sum())
print("Missing Prior_Visits_12m only:", df["Prior_Visits_12m"].isnull().sum())
print()
print("Missing Insurance AND Specialty:", 
      (df["Insurance_Type"].isnull() & df["Provider_Specialty"].isnull()).sum())
print("Missing Insurance AND Prior_Visits:", 
      (df["Insurance_Type"].isnull() & df["Prior_Visits_12m"].isnull()).sum())
print("Missing Specialty AND Prior_Visits:", 
      (df["Provider_Specialty"].isnull() & df["Prior_Visits_12m"].isnull()).sum())
print("Missing all three:", 
      (df["Insurance_Type"].isnull() & df["Provider_Specialty"].isnull() & df["Prior_Visits_12m"].isnull()).sum())
```
